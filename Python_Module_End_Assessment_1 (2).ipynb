##**Task 3 – Create SaaS Customer Class (OOP)**

```python
class SaaSCustomer:
    def __init__(self, customer_id, name, signup_date, plan, monthly_fee, usage_hours):
        self.customer_id = customer_id
        self.name = name
        self.signup_date = signup_date
        self.plan = plan
        self.monthly_fee = monthly_fee
        self.usage_hours = usage_hours
    def validate_data(self):

        try:
            self.monthly_fee = float(self.monthly_fee)
        except (ValueError, TypeError):
            self.monthly_fee = 0.0
        try:
            self.usage_hours = int(self.usage_hours)
        except (ValueError, TypeError):
            self.usage_hours = 0
        if self.name is None or str(self.name).strip() == "":
            self.name = "Unknown Customer"
        if self.plan is not None:
            self.plan = str(self.plan).strip().title()
        else:
            self.plan = "Unknown"
        if self.plan.lower() == "Pro":
            self.plan = "Pro"

    def display_customer(self):
        print("Customer Name:", self.name)

customer = SaaSCustomer(
    '109',
    'Dev',
    '2020-02-20',
    'Pro',
    '499.99',
    '180'
)

customer.validate_data()
customer.display_customer()

print(customer.monthly_fee)
print(customer.usage_hours)
print(customer.plan)
```

##**Task 4 – Data Cleaning Pipeline**

```python
from datetime import datetime

class Customer:

  saas_data = {
  "customer_id": [101, 102, 103, 104, 105, 106, 107,108],
  "name": ["Rahul Sharma", "Meena Nair", None, "A. Kumar", "David", "Priya", "Rahul Sharma","Rahul"],
  "signup_date": ["2023-01-10", "2023/02/15", "03-01-2023", "15-04-2023", "2023-05-10",
  "20230520", "2023-01-10" , "2023-01-10"],
  "plan": ["Pro", "Basic", "Pro", "Enterprise", "Pro", "", "Pro","Pro"],
  "monthly_fee": ["49.99", "19.99", "forty-nine", "99.99", "49.99", "49.99", "49.99","49.99"],
  "usage_hours": ["", "45", "88", None, "NULL", "200", "120", "120"]
  }

  def Standardize_date_formats(self,saas_data):
    self.saas_data=saas_data
    signup_date_new=[]

    date_formats = [
            "%Y-%m-%d",
            "%Y/%m/%d",
            "%m-%d-%Y",
            "%d-%m-%Y",
            "%Y%m%d"
        ]

    for date_string in saas_data["signup_date"]:

            formatted_date = None

            for date_format in date_formats:
                try:
                    formatted_date = datetime.strptime(
                        date_string, date_format
                    )
                    break

                except ValueError:
                    continue

            if formatted_date:
                formatted_date_new = formatted_date.strftime("%Y-%m-%d")
                signup_date_new.append(formatted_date_new)
            else:
                signup_date_new.append("Invalid Date")

    self.saas_data['signup_date']=signup_date_new

    return self.saas_data

  def handle_missing_values(self,saas_data):
    self.saas_data=saas_data

    for i in range(len(saas_data["customer_id"])):
      if saas_data['name'][i] =="" or saas_data['name'][i] is None:
        self.saas_data['name'][i]="Unknown Customer"
      else:
        self.saas_data['name'][i]=saas_data['name'][i]

    for i in range(len(saas_data["customer_id"])):
      if saas_data['usage_hours'][i] =="" or saas_data['usage_hours'][i] is None or str(saas_data['usage_hours'][i]).strip().lower() in ["", "null", "none"]:
        self.saas_data['usage_hours'][i]=0
      else:
        self.saas_data['usage_hours'][i]=saas_data['usage_hours'][i]

    for i in range(len(saas_data["monthly_fee"])):

      if all(isinstance(x, (int, float)) for x in saas_data["monthly_fee"]):
        self.saas_data["monthly_fee"][i] = saas_data["monthly_fee"][i]
      else:
        self.saas_data["monthly_fee"][i] = 'None'



    return self.saas_data

  def remove_duplicate_records(self,saas_data):
    self.saas_data = saas_data
    unique_names = set()
    unique_records = []

    for i in range(len(self.saas_data["customer_id"])):

        name = self.saas_data["name"][i]

        if name not in unique_names:
            unique_names.add(name)
            unique_records.append(i)

    print("Unique record indexes:")
    print(unique_records)

    for key in self.saas_data:
        self.saas_data[key] = [
            self.saas_data[key][i]
            for i in unique_records
        ]

    return self.saas_data

print(f"Date values are formatted")
for i in range(len(saas_data["customer_id"])):

    print(
        f"Date: {saas_data['signup_date'][i]} "
    )
print("Duplicate customer names removed.")
for i in range(len(saas_data["customer_id"])):
        print(
        f"ID: {saas_data['customer_id'][i]} | "
        f"Name: {saas_data['name'][i]} | "
        f"Date: {saas_data['signup_date'][i]} | "
        f"Plan: {saas_data['plan'][i]} | "
        f"Fee: {saas_data['monthly_fee'][i]} | "
        f"Usage: {saas_data['usage_hours'][i]}"
    )
print(f'After Handling missing values')
for i in range(len(saas_data["customer_id"])):

    print(
        f"ID: {saas_data['customer_id'][i]} | "
        f"Name: {saas_data['name'][i]} | "
        f"Date: {saas_data['signup_date'][i]} | "
        f"Plan: {saas_data['plan'][i]} | "
        f"Fee: {saas_data['monthly_fee'][i]} | "
        f"Usage: {saas_data['usage_hours'][i]}"
    )
```

##**Task 5 – Summary Statistics and Insights**

```python
saas_data = {
  "customer_id": [101, 102, 103, 104, 105, 106, 107],
  "name": ["Rahul Sharma", "Meena Nair", None, "A. Kumar", "David", "Priya", "Rahul Sharma"],
  "signup_date": ["2023-01-10", "2023/02/15", "03-01-2023", "15-04-2023", "2023-05-10",
  "20230520", "2023-01-10"],
  "plan": ["Pro", "Basic", "Pro", "Enterprise", "Pro", "", "Pro"],
  "monthly_fee": ["49.99", "19.99", "forty-nine", "99.99", "49.99", "49.99", "49.99"],
  "usage_hours": ["120", "45", "88", None, "NULL", "200", "120"]
  }

total_customers = len(saas_data["customer_id"])
fees = []
for fee in saas_data["monthly_fee"]:
    try:
        fees.append(float(fee))
    except ValueError:
        pass

average_fee = sum(fees) / len(fees)

usage_hours = []
for hours in saas_data["usage_hours"]:
    try:
        usage_hours.append(int(hours))
    except (ValueError, TypeError):
        usage_hours.append(0)

total_usage = sum(usage_hours)

plan_count = {}

for plan in saas_data["plan"]:
    if plan == "":
        plan = "Unknown"

    if plan in plan_count:
        plan_count[plan] += 1
    else:
        plan_count[plan] = 1


print("Total number of customers:", total_customers)
print("Average monthly fee:", round(average_fee, 2))
print("Total usage hours:", total_usage)
print("Number of customers per plan:", plan_count)
```
